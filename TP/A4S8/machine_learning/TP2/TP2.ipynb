```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDClassifier
from sklearn.neural_network import MLPClassifier



import matplotlib.pyplot as plt
import numpy as np
```

```python
data_1 = np.loadtxt('dataset1.dat')
data_2 = np.loadtxt('dataset2.txt')
data_3 = np.loadtxt('dataset3.txt')
```

```python
X1 = data_1[:,[0, 1]]
y1 = data_1[:,2]

print(len(np.unique(y1)))
```

```python
X2 = data_2[:,[0, 1]]
y2 = data_2[:,2]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X1, y1, train_size=0.7)
```

```python
def frontiere(model, A, B):
    couleur = []

    for i in y_train:
        couleur.append(int(i))

    x_min, x_max = X_train[:, 0].min()*1.1, X_train[:, 0].max()*1.1
    y_min, y_max = X_train[:, 1].min()*1.1, X_train[:, 1].max()*1.1
    x_h = (x_max - x_min)/50
    y_h = (y_max - y_min)/50
    xx, yy = np.meshgrid(np.arange(x_min, x_max, x_h),
    np.arange(y_min, y_max, y_h))
    Y = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Y = Y.reshape(xx.shape)
    #afficher les frontières/données d'apprentissage
    plt.contourf(xx, yy, Y, cmap=plt.cm.Paired, alpha=0.8)
    colors = np.array([x for x in "rgbcmyk"])
    plt.scatter(X_train[:, 0], X_train[:, 1], cmap=plt.cm.Paired, color=colors[couleur])
    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())
    plt.title(f'{A} : {B}')
    plt.show()
```

```python
def experience(A, B):
    clf1=SGDClassifier(loss='squared_epsilon_insensitive', eta0=A, max_iter=B, learning_rate='constant')
    clf1.fit(X_train, y_train)

    print(f'coef : {clf1.coef_}')
    print(f'sueil : {clf1.intercept_}')
    score = clf1.score(X_test, y_test)

    frontiere(clf1, A, B)
    return score
```

```python
def tableau(a, b):
    tab = []
    for i in range(5):
        tab.append(experience(a, b))


    print(f'{a} : {b}')
    print(np.mean(tab))
    print(np.std(tab))
```
