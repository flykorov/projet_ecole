```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import accuracy_score
```

## Chargement des données

1. 

```python
BC = load_breast_cancer()

X = BC.data
y = BC.target
```

```python
nb_desc = X.shape[1]
nb_ex = X.shape[0]
nb_classe = len(np.unique(y))
```

```python
print(f'Il y a {nb_desc} descriptions.')
print(f'Il y a {nb_ex} d\'exemples.')
print(f'Il y a {nb_classe} classes.')
```

2. 

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

3.

```python
n = NearestNeighbors(n_neighbors=1)
n.fit(X_train, y_train)

k = n.kneighbors(X_test, return_distance=False)

y_pred = y_train[k]

acc = accuracy_score(y_pred, y_test)

print(f'accuracy score de {acc}')
```

## Filter methodds

```python
X_10 = SelectKBest(chi2, k=10).fit(X, y)

print(X_10.get_support(indices=True))
X_10.transform(X)
print(X_10.scores_)
```
