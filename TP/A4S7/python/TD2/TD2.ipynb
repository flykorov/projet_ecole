# Production problem 

Problème de Production

A manufacturer produces 2 types of strawberry yogurts A and B from Strawberry, Milk and Sugar. Each yogurt must respect the following proportions of raw materials.

<table>
<tr> <td> </td> <td> A </td> <td> B </td> </tr>
<tr> <td> Strawberry </td> <td> 2 </td> <td> 1 </td> </tr>
<tr> <td> Milk </td> <td> 1 </td> <td> 2 </td> </tr>
<tr> <td> Sugar </td> <td> 0 </td> <td> 1 </td> </tr>
</table>
We have 800 kg of strawberries, 700 kg of milk and 300 kg of sugar.
The sale of 1 kg of yoghurts A and B brings in € 4 and € 5 respectively.

The manufacturer seeks to maximize his profit.

---

Un  fabricant  produit  2  types  de  yaourts  à  la  fraise  A  et  B  à partir de  Fraise,  de  Lait  et  de  Sucre.  Chaque  yaourt  doit  respecter  les proportions  suivantes  de  matières  premières.

On  dispose  de  800  Kg  de  Fraises,  700  Kg  de  Lait  et  300  Kg  de sucre.
La  vente  de  1  Kg  de  yaourts  A  et  B  rapporte  respectivement  4€ et 5€.

Le  fabricant  cherche  à  maximiser  son  proﬁt.

## What quantities can we work with?

Sur  quelles  quantités  peut-on  travailler?

Only non-constant values: the quantities of yoghurts A and B produced

* We talk about variables

---

## What are we trying to optimize?

Que cherche-t-on à optimiser ?

The profit $ z $
* Calculated from $ x_A $ and $ x_B $
* We speak of objective function
* $ z = 4x_A + 5x_B $

---

Le profit $z$
* Calculé à partir de $x_A$ et $x_B$
* On  parle  de  fonction  objectif
* $ z =  4x_A + 5x_B$

## What are the constraints of the problem?

Quelles sont les contraintes du problème ?

First constraint: 800 Kg of strawberries available
* the quantity used depends on the production: $ 2x_A + x_B $
* $ 2x_A + x_B \leq $ 800 (strawberries)
* $ x_A + 2x_B \leq $ 700 (milk)
* $ x_B \leq $ 300 (sugar)
* $ x_A, x_B \geq 0 $ (positivity)

---

Première  contrainte  :  800  Kg  de  fraises  disponibles
* la  quantité  utilisée  dépend  de  la  production: $2x_A + x_B$
* $2x_A + x_B \leq 800$ (fraises)
* $x_A + 2x_B \leq 700$ (lait)
* $x_B \leq 300$ (sucre)
* $x_A, x_B \geq 0$ (positivité)

## Linear Program

Programme Linéaire

$$\begin{array}{r c l}
\max z=&4x_A&+5x_B\\
&2x_A&+x_B&\leq800\\
&x_A&+2x_B&\leq700\\
&&x_B&\leq300\\
&x_A,&x_B&\geq0
\end{array}
$$

# SCiPy

SciPy is easy to configure. Once you have it installed, you will have everything you need to get started. Its scipy.optimize sub-package can be used for linear and non-linear optimization.

To define and fix optimization issues with SciPy, you need to import `scipy.optimize.linprog ()`:

---

SciPy est simple à configurer. Une fois que vous l'aurez installé, vous aurez tout ce dont vous avez besoin pour commencer. Son sous-package scipy.optimize peut être utilisé pour l'optimisation linéaire et non linéaire.

Pour définir et résoudre les problèmes d'optimisation avec SciPy, vous devez importer `scipy.optimize.linprog()` :

```python
from scipy.optimize import linprog
```

Now that you've imported `linprog ()`, you can start optimizing.

`linprog ()` only solves minimization problems (no maximization) and does not allow inequality constraints with sign greater than or equal to (≥). To work around these issues, you should modify your problem before you start the optimization:

1. Instead of maximizing $ z = 4x_A + 5x_B $, you can minimize its negative
$ −z = −4x_A − 5x_B $.
2. Likewise, if there are $ \geq $, they must be inverted into $ \leq $.

This system is equivalent to the original and will have the same solution. The only reason to apply these changes is to overcome SciPy's limitations related to the formulation of the problem.

---

Maintenant que vous avez importé `linprog ()`, vous pouvez commencer à optimiser.

`linprog ()` ne résout que les problèmes de minimisation (pas de maximisation) et n'autorise pas les contraintes d'inégalité avec le signe supérieur ou égal au (≥). Pour contourner ces problèmes, vous devez modifier votre problème avant de commencer l'optimisation :

1. Au lieu de maximiser  $z = 4x_A+5x_B$, vous pouvez minimiser son négatif
$−z=−4x_A−5x_B$.
2. De même, s'il y a des $\geq$ il faut les inverser en $\leq$.

Ce système est équivalent à l'original et aura la même solution. La seule raison d'appliquer ces changements est de surmonter les limitations de SciPy liées à la formulation du problème.

## Defining input values

Définition des valeurs d'entrée

```python
obj = [-4, -5]
#       ─┬  ─┬
#        │   └┤ Coefficient de xA
#        └────┤ Coefficient de xB

lhs_ineq = [[ 2,  1],  # fraises  left side
            [ 1,  2],  # lait left side
            [ 0,  1]]  # sucre left side

rhs_ineq = [800,  # fraises constraint right side
            700,  # lait constraint right side
            300]  # sucre constraint right side
```

You place the values from the above system into the appropriate NumPy lists, tuples, or arrays:

* obj contains the coefficients of the objective function.
* lhs_ineq contains the left coefficients of the inequality constraints.
* rhs_ineq contains the coefficients on the right side of the inequality constraints.
* lhs_eq contains the coefficients of the left side of the equality constraint.
* rhs_eq contains the coefficients on the right side of the equality constraint.

---

Vous placez les valeurs du système ci-dessus dans les listes, tuples ou tableaux NumPy appropriés :

* obj contient les coefficients de la fonction objectif.
* lhs_ineq contient les coefficients de gauche des contraintes d'inégalité.
* rhs_ineq contient les coefficients du côté droit des contraintes d'inégalité.
* lhs_eq contient les coefficients du côté gauche de la contrainte d'égalité.
* rhs_eq contient les coefficients du côté droit de la contrainte d'égalité.

## Remark

Remarque


Please pay attention to the order of rows and columns!

The order of the coefficients of the objective function and the left sides of the constraints must match. Each column corresponds to a single decision variable.

---

Veuillez faire attention à l'ordre des lignes et des colonnes !

## Definition of the limits

Définition des bornes

```python
bnd = [(0, float("inf")),  # Bornes de xA
       (0, float("inf"))]  # Bornes de xB
```

Cette instruction est redondante car linprog() prend ces limites (de zéro à l'infini positif) par défaut.

## Calling linprog and reading the results

Appel de linprog et lecture des résultats

```python
opt = linprog(c=obj, A_ub=lhs_ineq, b_ub=rhs_ineq,
              A_eq = None, b_eq = None, bounds=bnd,
              method="revised simplex")
opt
```

* The parameter c refers to the coefficients of the objective function.
* A_ub and b_ub are related to the coefficients of the left and right sides of the inequality constraints, respectively.
* Likewise, A_eq and b_eq refer to equality constraints.
* You can use bounds to provide lower and upper bounds for decision variables.

You can use the Parameter method to define the linear programming method you want to use. There are three options:

1. method = "interior-point" selects the interior-point method. This option is set by default.
2. method = "revised simplex" selects the revised two-phase simplex method.
3. method = "simplex" selects the inherited two-phase simplex method.

---

* Le paramètre c fait référence aux coefficients de la fonction objectif. 
* A_ub et b_ub sont liés aux coefficients des côtés gauche et droit des contraintes d'inégalité, respectivement. 
* De même, A_eq et b_eq font référence à des contraintes d'égalité. 
* Vous pouvez utiliser des limites pour fournir les limites inférieure et supérieure des variables de décision.

Vous pouvez utiliser la méthode des paramètres pour définir la méthode de programmation linéaire que vous souhaitez utiliser. Il y a trois options :

1.   method="interior-point" sélectionne la méthode interior-point. Cette option est définie par défaut.
2. method="revised simplex" sélectionne la méthode révisée du simplex à deux phases.
3. method="simplex" sélectionne la méthode simplex à deux phases héritée.

linprog () returns a data structure with these attributes:

* .con are the residuals of the equality constraints.
* .fun is the value of the objective function at the optimum (if found).
* .message is the status of the solution.
* .nit is the number of iterations required to complete the calculation.
* .slack corresponds to the values of the margin variables, or to the differences between the values of the left and right sides of the constraints.
* .status is an integer between 0 and 4 that indicates the state of the solution, such as 0 when the optimal solution has been found.
* .success is a boolean indicating whether the optimal solution has been found.
* .x is a NumPy array containing the optimal values of the decision variables.

# PuLP


PuLP allows you to choose solvers and formulate problems in a more natural way. The default solver used by PuLP is the COIN-OR Branch and Cut Solver (CBC). It’s connected to the COIN-OR Linear Programming Solver (CLP) for linear relaxations and the COIN-OR Cut Generator Library (CGL) for cuts generation.

---

PuLP vous permet de choisir des solveurs et de formuler des problèmes de manière plus naturelle. Le solveur par défaut utilisé par PuLP est le solveur COIN-OR Branch and Cut (CBC). Il est connecté au solveur de programmation linéaire COIN-OR (CLP) pour les relaxations linéaires et à la bibliothèque de générateurs de coupes COIN-OR (CGL) pour la génération de coupes.

PuLP has a more convenient linear programming API than SciPy. You don't need to mathematically modify your problem or use vectors and matrices. Everything is cleaner and less prone to errors.

As usual, you start by importing what you need:

---

PuLP a une API de programmation linéaire plus pratique que SciPy. Vous n'avez pas besoin de modifier mathématiquement votre problème ou d'utiliser des vecteurs et des matrices. Tout est plus propre et moins sujet aux erreurs.

Comme d'habitude, vous commencez par importer ce dont vous avez besoin :

## Create the model

Créer le modèle

```python
# Create the model
model = LpProblem(name="production", sense=LpMaximize)
```

## Initialize decision variables

Initialiser les variables de décision

```python
# Initialize the decision variables
xA = LpVariable(name="xA", lowBound=0)
xB = LpVariable(name="xB", lowBound=0)
```

## Add model constraints

Ajouter les constraintes du modèle


```python
# Add the constraints to the model
model += (2 * xA +    xB <= 800, "fraises_constraint")
model += (    xA + 2 *xB <= 700, "lait_constraint")
model += (            xB <= 300, "sucre_constraint")
```

## Addition of the objective function

Ajout de la fonction objectif

```python
# Add the objective function to the model
obj_func = 4* xA + 5 * xB
model += obj_func
```

## Model display and verification

Affichage du modèle et vérification

```python
model
```

## Solve the problem

Résoudre le problème



```python
# Solve the problem
status = model.solve()
```

## Reading the results

Lecture des résultats

```python
print(f"status: {model.status}, {LpStatus[model.status]}")
```

```python
print(f"objective: {model.objective.value()}")
```

```python
for var in model.variables():
  print(f"{var.name}: {var.value()}")
```
